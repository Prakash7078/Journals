```python
import pandas as pd
data=pd.read_excel('JMIS_Articles_new.xlsx')
data.head()
```

```python
data.drop('Volume_Issue',axis=1, inplace=True)
```

```python
print(data.info())
print(data.isnull().sum())
print(data.describe())
```

```python

df = data.fillna('NULL')
```

```python
print(data.describe())
```

```python
df = data.where(pd.notna(data), None)

```

```python
df = data.replace("", None)  # Replaces empty spaces with None
```

```python
df['Year'] = df['Month_Year'].apply(lambda x: x.split(',')[1].strip().split()[0])
df['Year'] = df['Year'].astype(int)
```

```python
df.head()
```

```python
df.rename(columns={'Month_Year': 'Volume_Issue'}, inplace=True)
```

```python
#to drop the unnecessary rows in the data
df_filtered = df[df['Year'] > 2005]
```

```python
df_filtered.to_excel('JMIS.xlsx',index=False)
```
